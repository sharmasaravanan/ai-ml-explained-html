```python
import gradio as gr

def energy_consumption_calculator(area, usage_hours, appliance_power):
  """Calculates estimated energy consumption."""
  # Simple estimation: power (kW) * hours (h)
  # Assuming appliance_power is in Watts, convert to kW
  energy_kwh = (appliance_power / 1000) * usage_hours
  return f"Estimated daily energy consumption: {energy_kwh:.2f} kWh"

# Create Gradio interface
iface = gr.Interface(
    fn=energy_consumption_calculator,
    inputs=[
        gr.Number(label="Area (sq ft or sq m)"),
        gr.Number(label="Daily Usage Hours"),
        gr.Number(label="Appliance Power (Watts)")
    ],
    outputs="text",
    title="Simple Energy Consumption Estimator",
    description="Estimate daily energy consumption based on area, usage hours, and appliance power."
)

# Launch the interface
iface.launch(debug=True)
```

```python
from bs4 import BeautifulSoup # remove html tags
import re # remove special characters
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords # remove stopwords
```

```python
words = stopwords.words("english")
```

```python
words
```

```python
def cleaning(raw):
    text = BeautifulSoup(raw, "html.parser").get_text()
    # print(text)
    # print()
    # Remove non-letters
    letters_only = re.sub("[^a-zA-Z]", " ", text)
    # print(letters_only)
    # print()
    # lower case
    lower_text = letters_only.lower()
    # print(lower_text)
    # print()
    # split the sentence intom words
    word = lower_text.split()
    # print(word)
    # print()
    # remove stopwords
    meaningful = [w for w in word if w not in words]
    # print(meaningful)
    # print()
    # join the words back into one string separated by space
    return " ".join(meaningful)
```
